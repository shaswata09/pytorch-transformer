## {'id': '0', 'translation': {'en': 'Source: Project Gutenberg', 'it': 'Source: www.liberliber.it/Audiobook available here'}}

```python
translation_list = []

en_list = ds_df['english_caption'].to_list()
bn_list = ds_df['bengali_caption'].to_list()

size = len(en_list)

if len(en_list) == len(bn_list):
    for i in range(size):
        tmp = {
            'en': en_list[i],
            'bn': bn_list[i]
        }
        translation_list.append(tmp)
```

```python
id_list = list(range(size))
```

```python
ds_dict = {
    'id': id_list,
    'translation': translation_list
}
```

```python
ds_df = pd.DataFrame(ds_dict)
```
